```python
import pandas as pd

# Load the S&P 500 historical data
sp500_data = pd.read_csv(r"C:\Users\jvlei\Desktop\TCC-pair-trading\data_bases\sp500_with_periods.csv")

# Display basic info
print("Data shape:", sp500_data.shape)
print("Columns:", sp500_data.columns[:10])  # First 10 columns
print("Unique periods:", sp500_data['periodo'].unique())

# Group by period and count non-null values for each company column
company_columns = sp500_data.columns[8:]  # Skip the first 8 metadata columns
composition = sp500_data.groupby('periodo')[company_columns].apply(lambda x: x.notna().sum())

print("Composition matrix shape:", composition.shape)
print("Sample composition for first period:")
print(composition.iloc[0].head(10))  # First 10 companies for first period
```

```python
# Count number of companies per period (columns with at least 1 non-null value)
companies_per_period = (composition > 0).sum(axis=1)
print("Number of companies per period:")
print(companies_per_period)

# Save the composition matrix to CSV
composition.to_csv('sp500_composition_by_period_1990_2014.csv')
print("Composition matrix saved to 'sp500_composition_by_period_1990_2014.csv'")

# Also save the company counts
companies_per_period.to_csv('companies_per_period.csv')
print("Company counts saved to 'companies_per_period.csv'")
```

```python
import pandas as pd
import yfinance as yf
from datetime import datetime, timedelta

# Load the companies list
with open('companies_2014_S2.txt', 'r') as f:
    companies = [line.strip() for line in f]

# Filter potential tickers (3-5 characters, letters and underscore only)
potential_tickers = [c for c in companies if 3 <= len(c) <= 5 and all(ch.isalpha() or ch == '_' for ch in c)]

print(f"Total companies: {len(companies)}")
print(f"Potential tickers: {len(potential_tickers)}")
print("Sample potential tickers:", potential_tickers[:10])

# Test fetching for a few
test_tickers = potential_tickers[:5]  # First 5
yahoo_data = {}

for ticker in test_tickers:
    try:
        data = yf.download(ticker, start='2014-07-01', end='2025-07-01')  # Up to June 2025
        if not data.empty:
            yahoo_data[ticker] = data
            print(f"{ticker}: SUCCESS - {len(data)} days")
        else:
            print(f"{ticker}: NO DATA")
    except Exception as e:
        print(f"{ticker}: ERROR - {str(e)}")

# Save one example
if yahoo_data:
    example_ticker = list(yahoo_data.keys())[0]
    yahoo_data[example_ticker].to_csv(f'{example_ticker}_yahoo_data.csv')
    print(f"Example data saved for {example_ticker}")
```

```python
# Fetch data for all potential tickers
yahoo_available = {}
yahoo_prices = {}

for i, ticker in enumerate(potential_tickers):
    try:
        data = yf.download(ticker, start='2014-07-01', end='2025-07-01', progress=False)
        if not data.empty and len(data) > 100:  # At least some data
            yahoo_available[ticker] = 'AVAILABLE'
            yahoo_prices[ticker] = data['Adj Close']  # Use adjusted close
            print(f"{i+1}/{len(potential_tickers)}: {ticker} - SUCCESS ({len(data)} days)")
        else:
            yahoo_available[ticker] = 'NO_DATA'
            print(f"{i+1}/{len(potential_tickers)}: {ticker} - NO DATA")
    except Exception as e:
        error_msg = str(e).lower()
        if 'delisted' in error_msg or 'no price data' in error_msg or 'no timezone' in error_msg:
            yahoo_available[ticker] = 'DELISTED'
            print(f"{i+1}/{len(potential_tickers)}: {ticker} - DELISTED")
        else:
            yahoo_available[ticker] = f'ERROR: {str(e)[:50]}'
            print(f"{i+1}/{len(potential_tickers)}: {ticker} - ERROR")

# Save availability report
pd.DataFrame(list(yahoo_available.items()), columns=['Ticker', 'Status']).to_csv('yahoo_availability_report.csv', index=False)
print("Availability report saved.")

# Count successful
successful_tickers = [t for t, s in yahoo_available.items() if s == 'AVAILABLE']
print(f"Successful tickers: {len(successful_tickers)} out of {len(potential_tickers)}")

# Now, to extend the data, we need to create new periods
# Define periods from 2015/S1 to 2025/S1
periods = []
for year in range(2015, 2026):
    for sem in [1, 2]:
        if year == 2025 and sem == 2:
            continue  # Only up to S1 2025
        start_month = 1 if sem == 1 else 7
        end_month = 6 if sem == 1 else 12
        start_date = f"{year}-{start_month:02d}-01"
        end_date = f"{year}-{end_month:02d}-31"
        periodo = f"{year}/S{sem}"
        periods.append({
            'ano': year,
            'semestre': sem,
            'periodo': periodo,
            'inicio_semestre': start_date,
            'fim_semestre': end_date
        })

periods_df = pd.DataFrame(periods)
print(f"New periods to add: {len(periods_df)}")
print(periods_df.head())
```
